````python
# 05_llm_vs_rule_guardrails.ipynb
# Claim (a), extended with a MODEL-CAPABILITY axis: the LLM guardrail agent is
# not a reproduction of the fixed-ratio rule baseline, AND its behaviour changes
# systematically with model capability (weak -> mid -> strong).
#
# Three model tiers are read from environment variables (see your .env):
#     LLM_MODEL_WEAK   (e.g. gpt-4o-mini)
#     LLM_MODEL_MID    (e.g. gpt-5.4-mini)
#     LLM_MODEL_STRONG (e.g. gpt-5.4)
# and an OPENAI_API_KEY. Load them however you prefer, e.g.:
#     from dotenv import load_dotenv; load_dotenv()
#
# API-style routing: GPT-5.x models are called through the Responses API
# (client.responses.create); older chat models through Chat Completions. Both
# paths request strict JSON and are wrapped so a tier that is unavailable is
# skipped rather than aborting the run.
#
# Only Layer-1 safety floors are enforced in code (identically for every tier
# and for the rule baseline), so any divergence is confined to the delegated
# Layer-2 ranges -- exactly where model capability should show up.
#
# Three axes of comparison
#   Axis 1 (divergence)  : interval IoU vs the rule baseline (lower = more
#                          patient-specific, less rule-like).
#   Axis 2 (recourse)    : feasibility, achieved class, changed-variable count
#                          when DiCE runs under each tier's guardrail.
#   Axis 3 (safety)      : how often the RAW (pre-Layer-1) LLM range would have
#                          breached an absolute safety floor -- i.e. how much the
#                          code-level guardrail had to correct each tier.
#
# Inputs  (../data/): df_final.pkl, agent_config.pkl, model_*.pkl, val_*.pkl
# Outputs (../results/tables/): llm_tier_divergence.csv,
#                     table_tier_divergence.tex
# Figures (../results/figures/): fig_tier_iou, fig_tier_safety_breach

import os
import json
import joblib
import numpy as np
import pandas as pd
import openai

import guardrail_core as gc
import experiment_core as ec

# --- load .env (key + model tiers) ---
try:
    from dotenv import load_dotenv, find_dotenv
    load_dotenv(find_dotenv())
except Exception:
    pass

# --- plotting style: grayscale, dpi 600, PNG+PDF, no captions ---
import matplotlib as mpl
import matplotlib.pyplot as plt
import seaborn as sns
FIG_DIR = os.path.join("..", "results", "figures")
TAB_DIR = os.path.join("..", "results", "tables")
os.makedirs(FIG_DIR, exist_ok=True); os.makedirs(TAB_DIR, exist_ok=True)
sns.set_theme(style="white", context="paper"); sns.set_palette("Greys")
mpl.rcParams.update({"savefig.dpi": 600, "font.family": "DejaVu Sans",
    "font.size": 10, "axes.edgecolor": "#000000", "axes.linewidth": 0.8,
    "axes.spines.top": False, "axes.spines.right": False, "axes.grid": False})
GREY_LEVELS = ["#BDBDBD", "#888888", "#3A3A3A"]   # weak / mid / strong
HATCHES     = ["//", "..", "xx"]
def save_fig(fig, name):
    for ext in ("png", "pdf"):
        pth = os.path.join(FIG_DIR, f"{name}.{ext}")
        fig.savefig(pth, dpi=600, bbox_inches="tight", format=ext); print("  saved ->", pth)

DATA = os.path.join("..", "data")
df_final = joblib.load(os.path.join(DATA, "df_final.pkl"))
ac = joblib.load(os.path.join(DATA, "agent_config.pkl"))
X = ac["X_features"]; T = ac["target_col"]; VARY = ac["vary_features"]

_key = os.environ.get("OPENAI_API_KEY", "")
client = openai.OpenAI(api_key=_key) if _key else None
print("key loaded:", bool(_key))

# Model tiers from environment (with sensible fallbacks).
TIERS = {
    "weak":   os.environ.get("LLM_MODEL_WEAK",   "gpt-4o-mini"),
    "mid":    os.environ.get("LLM_MODEL_MID",    "gpt-5.4-mini"),
    "strong": os.environ.get("LLM_MODEL_STRONG", "gpt-5.4"),
}
print("Model tiers:", TIERS)

# Variables delegated to Layer 2 (where LLM and rule may diverge).
DELEGATED = ["BMI", "WaistCirc", "Weight", "Energy_kcal",
             "Sodium_mg", "Carb_g", "Sugar_g", "Fat_g"]

# Absolute safety floors used for the Axis-3 breach check (mirror guardrail_core).
SAFETY_FLOOR = {
    "Energy_kcal": 500.0, "Sodium_mg": 800.0, "Protein_g": 30.0,
    "Potassium_mg": 500.0, "Carb_g": 30.0, "Fiber_g": 5.0,
}


def _build_prompt(patient_row, gender_name):
    info = {k: float(patient_row[k]) for k in X}
    return f"""
You are a high-precision Clinical Data Strategist for a health insurer.
Respond ONLY in valid JSON using the exact variable names given.

[Patient ({gender_name})]
{json.dumps(info, ensure_ascii=False, indent=1)}

[Variables]
{", ".join(X)}

[Principles]
- Propose a permitted_range [min, max] per variable, tailored to THIS patient's
  baseline (NOT a fixed percentage template applied to everyone).
- Reductions only for BMI/WaistCirc/Weight; keep them mutually consistent.
- Cap Carb_g and Sugar_g at current when Sodium_mg is reduced.
- Keep Energy_kcal at or below current.
- Tailor the aggressiveness of each range to the patient's risk profile.

[Format - valid JSON only]
{{"reasoning": "...", "guardrail_ranges": {{"Var": [min, max]}}}}
"""


def _parse_json(text):
    text = text.strip()
    if text.startswith("```"):
        text = text.strip("`")
        text = text[text.find("{"):]
    return json.loads(text)


def call_model(model_name, prompt):
    """
    Route to the correct API. GPT-5.x -> Responses API; others -> Chat
    Completions. Returns the raw guardrail_ranges dict, or raises.
    """
    is_gpt5 = model_name.replace("-", "").lower().startswith("gpt5")
    if is_gpt5:
        # Responses API (recommended for GPT-5.x)
        resp = client.responses.create(
            model=model_name,
            input=[{"role": "system",
                    "content": "Respond strictly in valid JSON."},
                   {"role": "user", "content": prompt}],
        )
        # Responses API exposes aggregated text via output_text.
        text = getattr(resp, "output_text", None)
        if text is None:
            # fallback: walk the output structure
            chunks = []
            for item in getattr(resp, "output", []) or []:
                for c in getattr(item, "content", []) or []:
                    if getattr(c, "type", "") in ("output_text", "text"):
                        chunks.append(getattr(c, "text", ""))
            text = "".join(chunks)
        data = _parse_json(text)
    else:
        # Chat Completions with JSON mode
        resp = client.chat.completions.create(
            model=model_name,
            messages=[{"role": "system", "content": "Respond strictly in valid JSON."},
                      {"role": "user", "content": prompt}],
            response_format={"type": "json_object"},
        )
        data = json.loads(resp.choices[0].message.content)
    return data.get("guardrail_ranges", {})


def raw_safety_breaches(raw_ranges, cur):
    """
    Axis 3: count how many delegated variables the RAW (pre-Layer-1) LLM range
    would have pushed below an absolute safety floor. This is what the code-level
    guardrail has to fix; a higher count means the tier is less self-safe.
    """
    breaches = 0
    for feat, floor in SAFETY_FLOOR.items():
        if feat in raw_ranges:
            try:
                lo = float(raw_ranges[feat][0])
            except Exception:
                continue
            if lo < floor - 1e-6:
                breaches += 1
    # anthropometric ceiling breach: proposing an INCREASE (hi > current)
    for feat in ["BMI", "WaistCirc", "Weight"]:
        if feat in raw_ranges:
            try:
                hi = float(raw_ranges[feat][1])
            except Exception:
                continue
            if hi > cur[feat] + 1e-6:
                breaches += 1
    return breaches


# Full cohort: every model-confirmed Class 3 patient (no sampling).
def cohort_ids(model, val):
    Xv, yv = val["X_val"], val["y_val"]
    return Xv[(yv == 3) & (model.predict(Xv) == 3)].index.tolist()


RESULT_CSV = os.path.join(TAB_DIR, "llm_tier_divergence.csv")
FIELDNAMES = (["pid", "gender", "tier", "model", "mean_iou",
               "raw_safety_breaches", "feasible", "achieved_class",
               "n_changed_vars"] + [f"iou_{f}" for f in DELEGATED])


def _done_pairs():
    """Return set of (pid, gender) already fully done (all available tiers)."""
    if not os.path.exists(RESULT_CSV):
        return set(), pd.DataFrame(columns=FIELDNAMES)
    prev = pd.read_csv(RESULT_CSV)
    # a patient is 'done' if it has a row for every tier in TIERS
    counts = prev.groupby(["pid", "gender"])["tier"].nunique()
    done = set(counts[counts >= len(TIERS)].index.tolist())
    return done, prev


def _append_rows(rows):
    """Append rows to the CSV, writing a header only if the file is new."""
    import csv
    new = not os.path.exists(RESULT_CSV)
    with open(RESULT_CSV, "a", newline="", encoding="utf-8") as f:
        w = csv.DictWriter(f, fieldnames=FIELDNAMES)
        if new:
            w.writeheader()
        for r in rows:
            w.writerow(r)


def run(gender_code, gender_name, model_file, val_file, done_pairs):
    """
    Full-cohort run with incremental save. Each patient's rows are written to
    disk as soon as they are computed, and patients already present in the CSV
    (all tiers) are skipped, so an interrupted run resumes without re-billing.
    """
    model = joblib.load(os.path.join(DATA, model_file))
    val = joblib.load(os.path.join(DATA, val_file))
    df_stable = df_final[df_final["Sex"] == gender_code].copy().astype(float)
    exp = ec.make_dice(model, df_stable, X, T)
    idx = cohort_ids(model, val)
    todo = [pid for pid in idx if (pid, gender_name) not in done_pairs]
    print(f"{gender_name}: {len(idx)} confirmed Class 3, "
          f"{len(idx)-len(todo)} already done, {len(todo)} to process")

    for k, pid in enumerate(todo):
        q = val["X_val"].loc[[pid]][X]
        cur = {f: float(q.iloc[0][f]) for f in X}
        g_rule = gc.build_rule_guardrails(cur, X, "aggressive")

        patient_rows = []
        for tier, model_name in TIERS.items():
            try:
                raw = call_model(model_name, _build_prompt(q.iloc[0], gender_name))
            except Exception as e:
                print(f"  [{tier}:{model_name}] failed for {pid}: {e}")
                continue
            breaches = raw_safety_breaches(raw, cur)
            g_llm = gc.sanitize_llm_guardrails(raw, cur, X, "aggressive")
            div = gc.guardrail_divergence(g_llm, g_rule, cur, DELEGATED)
            r_llm = ec.eval_pre_injection(exp, q, df_stable, X, VARY, g_llm)
            row = {"pid": pid, "gender": gender_name, "tier": tier,
                   "model": model_name, "mean_iou": div["_mean_iou"],
                   "raw_safety_breaches": breaches,
                   "feasible": r_llm["feasible"],
                   "achieved_class": r_llm["achieved_class"],
                   "n_changed_vars": r_llm["n_changed_vars"]}
            for f in DELEGATED:
                row[f"iou_{f}"] = div[f]["iou"]
            patient_rows.append(row)
        # incremental save: persist this patient before moving on
        if patient_rows:
            _append_rows(patient_rows)
        if (k + 1) % 5 == 0:
            print(f"  {gender_name} {k+1}/{len(todo)} done "
                  f"(last pid {pid}, IoU={patient_rows[-1]['mean_iou']:.2f} "
                  f"if available)")


# ---------------------------------------------------------------------------
# Run on the FULL cohort with incremental save + resume.
# Re-running this cell continues where it left off; delete
# results/tables/llm_tier_divergence.csv to start over.
# ---------------------------------------------------------------------------
if client is None:
    print("OPENAI_API_KEY not set. Set the key and the LLM_MODEL_* env vars, "
          "then re-run. No API calls were made.")
    res = pd.DataFrame()
else:
    done_pairs, _ = _done_pairs()
    run(1.0, "Male",   "model_male.pkl",   "val_male.pkl",   done_pairs)
    run(2.0, "Female", "model_female.pkl", "val_female.pkl", done_pairs)
    res = pd.read_csv(RESULT_CSV)
    # Resume can append duplicate rows for a re-processed patient; keep the last
    # occurrence per (pid, gender, tier) and rewrite the clean CSV.
    before = len(res)
    res = res.drop_duplicates(subset=["pid", "gender", "tier"], keep="last")
    if len(res) != before:
        res.to_csv(RESULT_CSV, index=False)
        print(f"  de-duplicated {before-len(res)} resumed row(s)")
    print(f"\nsaved -> llm_tier_divergence.csv ({len(res)} rows, "
          f"{res['pid'].nunique()} patients)")

# ---------------------------------------------------------------------------
# Analysis + figures (only if we have results)
# ---------------------------------------------------------------------------
if len(res):
    TIER_ORDER = ["weak", "mid", "strong"]
    present = [t for t in TIER_ORDER if t in res["tier"].unique()]

    print("\n=== Axis 1: divergence from rule (mean IoU; lower = less rule-like) ===")
    print(res.groupby("tier")["mean_iou"].agg(["mean", "std"]).reindex(present).round(3).to_string())
    print("\n=== Axis 3: raw safety-floor breaches per patient (higher = less self-safe) ===")
    print(res.groupby("tier")["raw_safety_breaches"].agg(["mean", "max"]).reindex(present).round(3).to_string())
    print("\n=== Axis 2: recourse under each tier's guardrail ===")
    feas = res[res["feasible"]]
    print(res.groupby("tier")["feasible"].mean().reindex(present).round(3).to_string())
    print(feas.groupby("tier")["n_changed_vars"].mean().reindex(present).round(2).to_string())

    # Figure: mean IoU by tier (grayscale)
    means = res.groupby("tier")["mean_iou"].mean().reindex(present)
    fig, ax = plt.subplots(figsize=(5.5, 4))
    xs = np.arange(len(present))
    for i, t in enumerate(present):
        ax.bar(i, means[t], color=GREY_LEVELS[i], hatch=HATCHES[i],
               edgecolor="white", linewidth=0.6)
        ax.text(i, means[t] + 0.01, f"{means[t]:.2f}", ha="center", va="bottom", fontsize=9)
    ax.axhline(1.0, color="#1A1A1A", ls=(0, (2, 2)), lw=0.8)
    ax.set_xticks(xs); ax.set_xticklabels([f"{t}\n{TIERS[t]}" for t in present], fontsize=8.5)
    ax.set_ylabel("Mean interval IoU vs rule", fontsize=10)
    ax.set_ylim(0, 1.05)
    plt.tight_layout(); save_fig(fig, "fig_tier_iou"); plt.show()

    # Figure: raw safety breaches by tier (grayscale)
    br = res.groupby("tier")["raw_safety_breaches"].mean().reindex(present)
    fig, ax = plt.subplots(figsize=(5.5, 4))
    for i, t in enumerate(present):
        ax.bar(i, br[t], color=GREY_LEVELS[i], hatch=HATCHES[i],
               edgecolor="white", linewidth=0.6)
        ax.text(i, br[t] + 0.02, f"{br[t]:.2f}", ha="center", va="bottom", fontsize=9)
    ax.set_xticks(np.arange(len(present)))
    ax.set_xticklabels([f"{t}\n{TIERS[t]}" for t in present], fontsize=8.5)
    ax.set_ylabel("Raw safety-floor breaches per patient", fontsize=10)
    plt.tight_layout(); save_fig(fig, "fig_tier_safety_breach"); plt.show()

    # LaTeX table
    lines = [r"\begin{table}[ht]", r"\centering",
             r"\caption{LLM guardrail behaviour across model tiers, on sampled "
             r"Class~3 patients. Interval IoU is measured against the fixed-ratio "
             r"rule baseline on delegated (Layer-2) variables (lower = more "
             r"patient-specific); raw safety breaches count how often the model's "
             r"pre-correction range would have crossed an absolute safety floor "
             r"(higher = more reliant on the code-level guardrail). Layer-1 floors "
             r"are enforced identically for every tier.}",
             r"\label{tab:tier_divergence}",
             r"\begin{tabular}{lccc}", r"\toprule",
             r"Tier (model) & Mean IoU vs rule & Raw safety breaches & Feasibility \\",
             r"\midrule"]
    for t in present:
        sub = res[res["tier"] == t]
        lines.append(
            f"  {t} ({TIERS[t]}) & {sub['mean_iou'].mean():.3f} "
            f"& {sub['raw_safety_breaches'].mean():.2f} "
            f"& {sub['feasible'].mean():.1%} \\\\")
    lines += [r"\bottomrule", r"\end{tabular}", r"\end{table}"]
    with open(os.path.join(TAB_DIR, "table_tier_divergence.tex"), "w") as f:
        f.write("\n".join(lines))
    print("\nsaved -> table_tier_divergence.tex")
print("\nDONE.")
````
